```python
# X - feature'lar (Pclass, Sex, Age), y - target (Survived)
X = df[["Pclass", "Sex", "Age"]].values
y = df["Survived"].values

n = len(df)
np.random.seed(42)          # har safar bir xil natija chiqishi uchun
idx = np.random.permutation(n)   # 0..n-1 sonlarini aralashtiramiz

split = int(n * 0.8)
train_idx, test_idx = idx[:split], idx[split:]

X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]

print("Train o'lchami:", X_train.shape)
print("Test o'lchami:", X_test.shape)
```

```python
n = len(df)
np.random.seed(42)
idx = np.random.permutation(n)
split = int(n * 0.8)
train_idx, test_idx = idx[:split], idx[split:]

X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]

print("Train:", X_train.shape, "Test:", X_test.shape)
print("NaN bormi:", np.isnan(X_train).sum())

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

w = np.zeros(X_train.shape[1])
b = 0.0

z = X_train @ w + b
p = sigmoid(z)
print("Birinchi 5 ta p:", p[:5])
```

```python
epsilon = 1e-15  # log(0) bo'lib qolmasligi uchun kichik himoya soni

def compute_loss(y_true, p):
    p_safe = np.clip(p, epsilon, 1 - epsilon)
    return -np.mean(y_true * np.log(p_safe) + (1 - y_true) * np.log(1 - p_safe))

loss = compute_loss(y_train, p)
print("Boshlang'ich loss:", loss)
```

```python
def compute_gradients(X, y_true, p):
    n = len(y_true)
    error = p - y_true          # har bir yo'lovchi uchun xato (712 ta son)
    dw = (1/n) * (X.T @ error)  # har bir feature uchun gradient (3 ta son)
    db = (1/n) * np.sum(error)  # bitta son
    return dw, db

dw, db = compute_gradients(X_train, y_train, p)
print("dw:", dw)
print("db:", db)
```

```python
w = np.zeros(X_train.shape[1])
b = 0.0
learning_rate = 0.1
epochs = 1000

for epoch in range(epochs):
    z = X_train @ w + b
    p = sigmoid(z)

    loss = compute_loss(y_train, p)
    dw, db = compute_gradients(X_train, y_train, p)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 100 == 0:
        print(f"Epoch {epoch:4d} | Loss: {loss:.4f}")

print("\nYakuniy w:", w)
print("Yakuniy b:", b)
```

```python
z_test = X_test @ w + b
p_test = sigmoid(z_test)
y_pred = (p_test >= 0.5).astype(int)

accuracy = np.mean(y_pred == y_test)
print("Test Accuracy:", accuracy)
```

```python
thresholds = np.linspace(0, 1, 100)
tpr_list = []
fpr_list = []

for t in thresholds:
    y_pred_t = (p_test >= t).astype(int)

    tp = np.sum((y_pred_t == 1) & (y_test == 1))
    fp = np.sum((y_pred_t == 1) & (y_test == 0))
    fn = np.sum((y_pred_t == 0) & (y_test == 1))
    tn = np.sum((y_pred_t == 0) & (y_test == 0))

    tpr = tp / (tp + fn) if (tp+fn) > 0 else 0   # True Positive Rate (Recall)
    fpr = fp / (fp + tn) if (fp+tn) > 0 else 0   # False Positive Rate

    tpr_list.append(tpr)
    fpr_list.append(fpr)

# AUC - egri chiziq ostidagi maydon (trapezoid usuli bilan taxminiy hisoblash)
auc = -np.trapz(tpr_list, fpr_list)
print("AUC:", auc)
```
